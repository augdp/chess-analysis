```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))


import chess  
from src.setup.stockfish import StockfishEngine
from src.compute.cp_eval import CPEvaluator
```

```python
fen = "r1bqkbnr/pp1pnBpp/2p5/5R2/P3P3/7R/1PPPK2P/1NBQ4 b kq - 0 13"
```

```python
board = chess.Board(fen)
```

```python
engine_path = os.path.join(os.path.abspath(".."), "engine/stockfish.exe")
engine = StockfishEngine(engine_path=engine_path, timeout=5.0)  
engine.start()
```

```python
cp_evaluator = CPEvaluator(engine_handle=engine, depth=20)
```

```python
eval_result = cp_evaluator.evaluate(board)
```

```python
print("Position FEN:", fen)  
print("Evaluation (cp):", eval_result["eval_cp"])  
print("Evaluation (mate):", eval_result["eval_mate"])
```
